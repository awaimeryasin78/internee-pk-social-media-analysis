```python
# Set the style for all our charts
sns.set_style("darkgrid")
plt.rcParams['figure.figsize'] = (12, 6)

# Create a figure with 4 charts inside it
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle("Internee.pk Social Media Engagement Analysis", fontsize=18, fontweight='bold', y=1.02)

# --- Chart 1: Average Engagement Rate by Platform ---
platform_eng = df.groupby('platform')['engagement_rate'].mean().round(2)
axes[0, 0].bar(platform_eng.index, platform_eng.values,
               color=['#E1306C', '#0077B5', '#1877F2'], edgecolor='white', linewidth=1.5)
axes[0, 0].set_title('Average Engagement Rate by Platform', fontsize=13, fontweight='bold')
axes[0, 0].set_xlabel('Platform')
axes[0, 0].set_ylabel('Engagement Rate (%)')
for i, v in enumerate(platform_eng.values):
    axes[0, 0].text(i, v + 0.3, f'{v}%', ha='center', fontweight='bold')

# --- Chart 2: Engagement Rate by Post Type ---
post_eng = df.groupby('post_type')['engagement_rate'].mean().round(2)
axes[0, 1].bar(post_eng.index, post_eng.values,
               color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'], edgecolor='white', linewidth=1.5)
axes[0, 1].set_title('Average Engagement Rate by Post Type', fontsize=13, fontweight='bold')
axes[0, 1].set_xlabel('Post Type')
axes[0, 1].set_ylabel('Engagement Rate (%)')
for i, v in enumerate(post_eng.values):
    axes[0, 1].text(i, v + 0.3, f'{v}%', ha='center', fontweight='bold')

# --- Chart 3: Best Days to Post ---
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
day_eng = df.groupby('day_of_week')['engagement_rate'].mean().reindex(day_order).round(2)
axes[1, 0].plot(day_order, day_eng.values, marker='o', color='#F7DC6F',
                linewidth=2.5, markersize=8, markerfacecolor='#E67E22')
axes[1, 0].set_title('Engagement Rate by Day of Week', fontsize=13, fontweight='bold')
axes[1, 0].set_xlabel('Day of Week')
axes[1, 0].set_ylabel('Engagement Rate (%)')
axes[1, 0].tick_params(axis='x', rotation=45)
for i, v in enumerate(day_eng.values):
    axes[1, 0].text(i, v + 0.5, f'{v}%', ha='center', fontsize=9)

# --- Chart 4: Likes by Platform (Box Plot) ---
df.boxplot(column='likes', by='platform', ax=axes[1, 1],
           patch_artist=True,
           boxprops=dict(facecolor='#AED6F1', color='navy'),
           medianprops=dict(color='red', linewidth=2))
axes[1, 1].set_title('Distribution of Likes by Platform', fontsize=13, fontweight='bold')
axes[1, 1].set_xlabel('Platform')
axes[1, 1].set_ylabel('Number of Likes')
plt.suptitle('')

plt.tight_layout()
plt.savefig('social_media_analysis.png', dpi=150, bbox_inches='tight')
plt.show()
print("Charts saved successfully!")
```

```python
# Check the shape of our data (rows, columns)
print("Data shape:", df.shape)

# Check for any missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# Make sure date column is in proper date format
df['date'] = pd.to_datetime(df['date'])

# Extract useful time features from the date
df['month'] = df['date'].dt.month
df['day_of_week'] = df['date'].dt.day_name()
df['month_name'] = df['date'].dt.month_name()

# Calculate Engagement Rate for each post
# Formula: (likes + comments + shares) / reach * 100
df['engagement_rate'] = ((df['likes'] + df['comments'] + df['shares']) / df['reach'] * 100).round(2)

# Show a summary of our cleaned data
print("\nCleaned Data Summary:")
print(df.describe().round(2))

print("\nFirst 5 rows with new columns:")
df.head()
```

```python
# Set a random seed so we get the same results every time
random.seed(42)
np.random.seed(42)

# Define our platforms and post types
platforms = ['Instagram', 'LinkedIn', 'Facebook']
post_types = ['Image', 'Video', 'Text', 'Story']

# Create a start date - we'll generate 6 months of data
start_date = datetime(2024, 1, 1)

# Empty list to store all our posts
all_posts = []

# Generate 100 posts per platform
for platform in platforms:
    for i in range(100):

        # Random date within 6 months
        post_date = start_date + timedelta(days=random.randint(0, 180))

        # Each platform has different typical engagement numbers
        if platform == 'Instagram':
            likes = random.randint(100, 5000)
            comments = random.randint(10, 300)
            shares = random.randint(5, 200)
            reach = random.randint(1000, 20000)

        elif platform == 'LinkedIn':
            likes = random.randint(50, 2000)
            comments = random.randint(5, 150)
            shares = random.randint(10, 300)
            reach = random.randint(500, 15000)

        elif platform == 'Facebook':
            likes = random.randint(30, 3000)
            comments = random.randint(5, 200)
            shares = random.randint(20, 400)
            reach = random.randint(800, 18000)

        # Add this post to our list
        all_posts.append({
            'platform': platform,
            'post_type': random.choice(post_types),
            'date': post_date,
            'likes': likes,
            'comments': comments,
            'shares': shares,
            'reach': reach
        })

# Convert list to a DataFrame (a table)
df = pd.DataFrame(all_posts)

# Show the first 5 rows
print("Total posts created:", len(df))
print("\nFirst 5 rows of our data:")
df.head()
```
